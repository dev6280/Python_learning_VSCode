```python
def removeDuplicates(arr):
    
    seen = set()

    idx = 0

    for i in range(len(arr)):
        if arr[i] not in seen:
            seen.add(arr[i])
            arr[idx] = arr[i]
            idx += 1
    return idx
```

```python
arr1=[10,9,8 ,9 ,8]
a = removeDuplicates(arr1)
for i in range(a):
    print(arr1[i], end = " ")
```
