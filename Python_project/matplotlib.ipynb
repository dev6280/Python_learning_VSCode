```python
import numpy as np
import matplotlib.pyplot as plt
x= np.arange(0,100)
y = x*2
z = x**2
```

```python
fig = plt.figure(figsize =(3,2))
ax = fig.add_axes([0,0,1,1])
ax.plot(x,y)
ax.set_title("TITLE")
ax.set_xlabel("x-axis")
ax.set_ylabel("y-axis")
```

```python
fig = plt.figure(figsize = (3,2))
ax1= fig.add_axes([0,0,1,1])
ax1.plot(x,y, color = "red", lw =.2, ls = ":" ,zorder =2, marker = "^", markersize = 2,markerfacecolor  = "blue", markeredgecolor = "black",label = "sales", alpha = .5)
ax1.legend(loc = 4)
ax2 = fig.add_axes([.2,0.5,.2,.2])
ax2.plot(x,y, color = "purple", lw =4)
```

```python
fig = plt.figure()
ax_1 = fig.add_axes([0,0,1,1])
ax_2 = fig.add_axes([.2,.5,.4,.4])
ax_1.plot(x,z)
ax_1.set_xlabel("xaxis")
ax_1.set_ylabel("zaxis")
ax1.set_title("Larger")

ax_2.plot(x,y)
ax_2.set_xlim(20,22)
ax_2.set_xlabel("xaxis")
ax_2.set_ylabel("yaxis")
ax_2.set_title("Smaller")
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1,2 , figsize  =(10,4))
ax[0].plot(x,y, color = "blue",lw = 3,ls = "--")

ax[1].plot(x,z, color = "red", lw =2)
```
